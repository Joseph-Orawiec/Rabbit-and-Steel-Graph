```python
# the point of the todo file was to manually human review each item pairing to determine if they synergized or not
# next few blocks were ran to read the file, print each items description and then save the progress to be picked up later
# The whole process didn't take too long, maybe 4 hours
# I also had a button that would remove all pairs with a certain item if i was confident it wouldn't synergize with anything else

def initialize():
    # open and read file, initialize the array
    arr = []
    file = open("todo.txt")
    lines = file.read().split('\n')
    for l in lines:
        arr.append(l.split("|"))
    file.close()
    return arr
```

```python
# This would be ran to stop and start the review process


finish = []
file = open("tdone.txt")
lines = file.read().split('\n')
for l in lines:
    finish.append(l.split("|"))
file.close()

if finish[-1] == ['']:
    finish.pop()

todo = initialize()
while len(todo) != 0:
    pair = todo[0]

    # print the pair
    print(pair[0])
    print(d[pair[0]])
    print()
    print(pair[1])
    print(d[pair[1]], flush=True)

# take user input
    u = input()
    while u == '':
        u = input()
    if u == "stop": # stop and save
        break
    if u == "s":
        # confident there's no synergies, the FIRST ITEM
        item = pair[0]
        for i in range(len(todo) - 1, -1, -1):
            if ((todo[i][0] == item)):
                todo.pop(i)

    else: # type in any number to count it as a synergy (usually would type in 1 and 0 for no)
        if int(u) != 0:
            pair.append(u)
            finish.append(pair)
            # remove other instances
            for i in range(len(todo) - 1, 0, -1):
                if todo[i][1] == pair[0]:
                    todo.pop(i)
        todo.pop(0)
    print("---------")

# write what's left of the todo file
file = open("tdone.txt", "w")
for i in finish:
    if i == finish[-1]:
        file.write(i[0] + "|" + i[1] + "|" + i[2])
    else:
        file.write(i[0] + "|" + i[1] + "|" + i[2] +  "\n")
file.close()

# overwrite the todo file
file = open("todo.txt", "w")
for i in todo:
    if i == todo[-1]:
        file.write(i[0] + "|" + i[1])
    else:
        file.write(i[0] + "|" + i[1] + "\n")
file.close()
```

```python
from pyvis.network import Network



g = Network(height="1000", width="1850", filter_menu=True, notebook=False)

group = list(a.Set)
image = list(a.Path)
effect = list(a.Effect)


for i in range(len(items)):
    g.add_node(
        n_id = items[i],
        label = items[i],
        shape = "circularImage",
        title = effect[i],
        group = group[i],
        image = image[i],
    )

# color switch
for i in g.get_nodes():
    g.get_node(i)['color'] = colors[group[items.index(i)]]
```

```python
file = open("tdone.txt")
edges = []
lines = file.read().split('\n')
for l in lines:
    edges.append(l.split("|"))
file.close()

for i in edges:
    g.add_edge(
        source=i[0], 
        to=i[1],
        width = i[2])
    
options = {
  "physics": {
    "minVelocity": 0.75,
    "solver": "repulsion"
  }
}

g.set_options('''
const options = {
  "physics": {
    "forceAtlas2Based": {
      "springLength": 100
    },
    "minVelocity": 0.75,
    "solver": "forceAtlas2Based"
  }
}
''')

# g.show_buttons(filter_=["physics"])
g.show("Ras.html")
```

```python
edges = g.get_edges()
d = {}
for i in items:
    d[i] = 0

for i in edges:
    d[i["from"]] = d[i["from"]] + 1
    d[i["to"]] = d[i["to"]] + 1
```
